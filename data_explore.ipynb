```python
import graph_tool.all as gt
```

```python
g = gt.load_graph("orkut.gt")
```

```python
print("Vertices:", g.num_vertices())
print("Edges:", g.num_edges())
```

```python
print(g.vertex_properties)
print(g.edge_properties)
print(g.graph_properties)
```

```python
print("Graph:", g)
print("Vertices:", g.num_vertices())
print("Edges:", g.num_edges())

print("\nVertex properties:")
for name in g.vertex_properties:
    print(" ", name)

print("\nEdge properties:")
for name in g.edge_properties:
    print(" ", name)

print("\nFirst 10 edges:")
for i, e in enumerate(g.edges()):
    print(int(e.source()), "->", int(e.target()))

    if i == 9:
        break
```

```python
name = g.vertex_properties["name"]

for i in range(20):
    v = g.vertex(i)
    print(i, name[v])
```

```python
communities = g.vertex_properties["communities"]
for i in range(20):
    v = g.vertex(i)

    print(
        "Node:", i,
        "| Name:", g.vertex_properties["name"][v],
        "| Communities:", communities[v]
    )
```

```python
multi = []

for v in g.vertices():
    c = communities[v]

    if len(c) > 1:
        multi.append(int(v))

print("Nodes in multiple communities:", len(multi))
```

```python
count = 0

for v in g.vertices():
    if len(communities[v]) > 1:
        count += 1

print("Nodes in multiple communities:", count)
```
